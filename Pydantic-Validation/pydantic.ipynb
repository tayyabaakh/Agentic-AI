```python
# Model Validator

from pydantic import BaseModel,EmailStr,AnyUrl,Field,field_validator,model_validator
from typing import List,Dict,Optional,Annotated

class Employee(BaseModel):
    name:str=Field(max_length=10)
    age:int=Field(gt=0,lt=60)
    email:EmailStr
    LinkedInUrl:AnyUrl
    salary:float
    department:str
    married:bool=False
    skills:Optional[List[str]]=Field(default=None,max_length=5)
    contact_info:Dict[str,str]

    @field_validator('email')
    @classmethod
    def email_validator(cls,value):
        valid_domains=['io.com','ici.com']
        domain_name=value.split('@')[-1]
        if domain_name not in valid_domains:
            raise ValueError("Not a valid domain")
        return value

    @field_validator('name')
    @classmethod
    def transform_name(cls,value):
        return value.upper()        


    @model_validator(mode='after')
    def validate_emergency_contact(cls,model):
        if model.age > 50 and 'emergency' not in model.contact_info:
            raise ValueError('employees older than 60 must have an emergency contact')
        return model

employee_data={"name":"tayyaba","age":"56","email":"tayyaba@io.com","LinkedInUrl":"https://www.linkedin.com/in/tayyaba-akhter-704149290/","salary":"30000","department":"Agentic AI","contact_info":{"email":"tayyaba@gmail.com","phone":"033333333", "emergency": "987-654-321"}}

def add_employee(employee:Employee):
    print(f"Adding {employee.name} to employees db with age: {employee.age} and {employee.email} and LinkedInUrl {employee.LinkedInUrl} of department {employee.department} with skills {employee.skills} and is  married {employee.married}.Her contact info is {employee.contact_info} ")

employee_one=Employee(**employee_data)
add_employee(employee_one)
```

```python
# Computed Fields

from pydantic import BaseModel,EmailStr,AnyUrl,Field,field_validator,model_validator,computed_field
from typing import List,Dict,Optional,Annotated

class Employee(BaseModel):
    name:str=Field(max_length=10)
    age:int=Field(gt=0,lt=60)
    tax_amount:int
    email:EmailStr
    LinkedInUrl:AnyUrl
    salary:float
    department:str
    married:bool=False
    skills:Optional[List[str]]=Field(default=None,max_length=5)
    contact_info:Dict[str,str]


    @field_validator('email')
    @classmethod
    def email_validator(cls,value):
        valid_domains=['io.com','ici.com']
        domain_name=value.split('@')[-1]
        if domain_name not in valid_domains:
            raise ValueError("Not a valid domain")
        return value

    @field_validator('name')
    @classmethod
    def transform_name(cls,value):
        return value.upper()        


    @model_validator(mode='after')
    def validate_emergency_contact(cls,model):
        if model.age > 50 and 'emergency' not in model.contact_info:
            raise ValueError('employees older than 60 must have an emergency contact')
        return model

    @computed_field
    @property
    def net_salary(self)->float:
        net_salary=round(self.salary-self.tax_amount)  
        return net_salary   

employee_data={"name":"tayyaba","age":"56","tax_amount":1000,"email":"tayyaba@io.com","LinkedInUrl":"https://www.linkedin.com/in/tayyaba-akhter-704149290/","salary":"30000","department":"Agentic AI","contact_info":{"email":"tayyaba@gmail.com","phone":"033333333", "emergency": "987-654-321"}}

def add_employee(employee:Employee):
    print(f"Adding {employee.name} to employees db with age: {employee.age} and {employee.email} and LinkedInUrl {employee.LinkedInUrl} of department {employee.department} with skills {employee.skills} and is  married {employee.married}.Her contact info is {employee.contact_info} ")
    print("net_salary", employee.net_salary)

employee_one=Employee(**employee_data)
add_employee(employee_one)
```
